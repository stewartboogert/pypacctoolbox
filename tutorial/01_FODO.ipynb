```python
import pypacctoolbox
import sympy
```

```python
d = pypacctoolbox.elementLinear.drift("d",sympy.Symbol('L_d'))
qf = pypacctoolbox.elementLinear.quadrupoleThin("qf",1/sympy.Symbol('f',negative=False))
qd = pypacctoolbox.elementLinear.quadrupoleThin("qd",-2/sympy.Symbol('f',negative=False))
l = pypacctoolbox.elementLinear.line("l",[qd,d,qf,d,qd])
```
