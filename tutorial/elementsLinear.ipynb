```python
import pypacctoolbox
import sympy
```

## Drift

```python
d1 = pypacctoolbox.elementLinear.drift("d1")
d2 = pypacctoolbox.elementLinear.drift("d2")
d_line = pypacctoolbox.elementLinear.line("d_line",[d1,d2])
```

```python
d_line.matrix
```

## Thick quad

```python
qf = pypacctoolbox.elementLinear.quadrupole("qf")
```

```python
qf.matrixXYZ
```

```python
qd = pypacctoolbox.elementLinear.quadrupole("qd",
                                            sympy.Symbol('L_d'),
                                            sympy.Symbol('k1_d',negative=True))
qd.matrixXYZ
```

```python
l = pypacctoolbox.elementLinear.line("line",[qf,d1])
l.matrixXYZ
```

## Thin quad

```python
tqf = pypacctoolbox.elementLinear.quadrupoleThin("tqf")
tqf.matrix
```

## FODO lattice

```python
tqf = pypacctoolbox.elementLinear.quadrupoleThin("tqf",
                                                 1/sympy.Symbol('f',negative=False))
tqd = pypacctoolbox.elementLinear.quadrupoleThin("tqd",
                                                 -2/sympy.Symbol('f',negative=False))
```

```python
tqf.matrix
```

```python
tqd.matrix
```

```python
l = pypacctoolbox.elementLinear.line("line",[tqf,d1,tqd,d1,tqf])
```

```python
l.matrix.simplify()
```

```python
l.matrix
```
