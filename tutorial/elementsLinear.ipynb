```python
import pypacctoolbox
import sympy
```

```python
d = pypacctoolbox.elementLinear.drift("d1",sympy.Symbol("L"))
```

```python
d.matrixXYZ;
```

```python
qf = pypacctoolbox.elementLinear.quadrupole("qf",
                                            sympy.Symbol('L_f'),
                                            sympy.Symbol('k1_f',negative=False))
```

```python
qf.matrixXYZ;
```

```python
qd = pypacctoolbox.elementLinear.quadrupole("qd",
                                            sympy.Symbol('L_d'),
                                            sympy.Symbol('k1_d',negative=True))
```

```python
qd.matrixXYZ;
```

```python
l = pypacctoolbox.elementLinear.line("line",[qf,d])
l.length
l.matrixXYZ
```

```python
l
```
