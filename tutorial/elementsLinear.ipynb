```python
import pypacctoolbox
import sympy
```

## Drift

```python
d = pypacctoolbox.elementLinear.drift("d1")
```

```python
d.matrixXYZ
```

## Thick quad

```python
qf = pypacctoolbox.elementLinear.quadrupole("qf")
```

```python
qf.matrixXYZ
```

```python
qd = pypacctoolbox.elementLinear.quadrupole("qd",
                                            sympy.Symbol('L_d'),
                                            sympy.Symbol('k1_d',negative=True))
qd.matrixXYZ
```

```python
l = pypacctoolbox.elementLinear.line("line",[qf,d])
l.matrixXYZ
```

## Thin quad and FODO

```python
tqf = pypacctoolbox.elementLinear.quadrupoleThin("tqf")
tqf.matrix
```

```python
tqf = pypacctoolbox.elementLinear.quadrupoleThin("tqf",
                                                 1/sympy.Symbol('f',negative=False))
tqd = pypacctoolbox.elementLinear.quadrupoleThin("tqd",
                                                 -2/sympy.Symbol('f',negative=False))
```

```python
tqf.matrix
```

```python
tqd.matrix
```

```python
l = pypacctoolbox.elementLinear.line("line",[tqf,d,tqd,d,tqf])
```

```python
l.matrix.simplify()
```

```python
l.matrix
```
