# Multi variable Taylor expansion

Taylor expansion, with multi-index notation

$T({\bf x}) = \sum_{\alpha > 0} \frac{(\bf x - a)^\alpha}{\alpha!} (\partial^{\alpha} f)({\bf a})$

$ \alpha = (\alpha_1, \alpha_2, ... \alpha_n)$

$ ({\bf x - a})^\alpha = (x_1 - a_1)^{\alpha_1} (x_2 - a_2)^{\alpha_2}... (x_n - a_n)^{\alpha_n}$

$ \partial^{\alpha} = \partial^{\alpha_1}_1 \partial^{\alpha_2}_2 .., \partial^{\alpha_n}_n $

$ \alpha! = \alpha_1! \alpha_2! ... \alpha_n!$

# Taylor expansion equations of motion

$\frac{\partial x}{\partial s} = -\frac{\partial H}{\partial p}$

$\frac{\partial p}{\partial s} = \frac{\partial H}{\partial x}$
