```python
x,y,z,p_x,p_y,delta = d.canonical_variables
beta0, gamma0 = d.relativistic_variables
```

```python
sympy.diff(d.hamiltonian,p_x)
```

```python
d.hamiltonian.subs({x:0, p_x:0, y:0, p_y:0, z:0, delta:0}).subs({gamma0:1/sympy.sqrt(1-beta0**2)}).simplify()
```

```python
d.hamiltonian.diff(p_x).subs({x:0, p_y:0, y:0, p_y:0, z:0, delta:0}).subs({gamma0:1/sympy.sqrt(1-beta0**2)}).simplify().series(p_x,n=2)
```

```python
d.hamiltonian.diff(p_y).subs({x:0, y:0, p_x:0, z:0, delta:0}).subs({gamma0:1/sympy.sqrt(1-beta0**2)}).simplify().series(p_y,n=2)
```

```python
d.hamiltonian.diff(delta).subs({x:0, y:0, p_x:0, p_y:0, z:0}).subs({gamma0:1/sympy.sqrt(1-beta0**2)}).simplify().series(delta,n=2)
```
