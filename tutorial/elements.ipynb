# Quadrupole

```python
l = sympy.Symbol("l")
k1 = sympy.Symbol("k1")
q = pypacctoolbox.element.Quadrupole("q",1,k1)
```

```python
q.hamiltonian
```

```python
q.equations_of_motion()
```

```python
beta0, gamma0 = q.relativistic_variables
pypacctoolbox.math.MultiIndex.taylor(q.hamiltonian,q.canonical_variables,3).subs(gamma0, 1/sympy.sqrt(1-beta0**2)).simplify().expand()

```
