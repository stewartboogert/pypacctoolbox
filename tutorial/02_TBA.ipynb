```python
import pypacctoolbox
import sympy
```

```python
b1 = pypacctoolbox.elementLinear.bendSector("b1")
d1 = pypacctoolbox.elementLinear.drift("d1")
q1 = pypacctoolbox.elementLinear.quadrupoleThin("q1")
d2 = pypacctoolbox.elementLinear.drift("d2")
q2 = pypacctoolbox.elementLinear.quadrupoleThin("q2")
d3 = pypacctoolbox.elementLinear.drift("d3")
b2 = pypacctoolbox.elementLinear.bendSector("b2")

q1_k1 = q1.k1
q2_k1 = q2.k1
beta = b1.beta
gamma = b1.gamma
```

```python
l = pypacctoolbox.elementLinear.line("l",[b1,d1,q1,d2,q2,d3,b2])
```

```python
d = l.matrix[4,5].subs(gamma,1).subs(beta,1)
```

```python
s = sympy.solve(sympy.Eq(d,0), q1_k1, q2_k1)[0][0]
print(s)
```
