# Multi-index

```python
import pypacctoolbox
```

```python
i1 = pypacctoolbox.math.MultiIndex([2,2,2])
i2 = pypacctoolbox.math.MultiIndex([3,2,2])
```

```python
i1.factorial()
```

```python
i1+i2
```

```python
i2-i1
```

```python
i1.norm()
```

```python
i1.iterationCount()
```

```python
for i in i2 :
    print(i+i)
```

```python
i1.multinomial(0)
```

```python
pypacctoolbox.math.MultiIndex.binominal(i1,i2)
```
